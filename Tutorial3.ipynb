```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/movies.csv')
display(df.head(10))
```

```python
print("Rows and columns:",df.shape)
print("Duplicate movies:", df.duplicated().sum())

# The dataset has 2000 samples which is a good amount of samples and has no duplicate movies.
```

```python
print("All columns:",df.columns)
print("")
print("Number of features:",len(df.columns)-1)

# The number of features is 12 (excluding the target), which is a suitable amount of features.
```

```python
print("Total samples:",len(df))

# The data set has 2000 samples which is enough samples.
```

```python
print("Data types for each column:")
print(df.dtypes)
print()
print("Data types present:")
print(df.dtypes.unique())

# The data set has 3 data types: object, float64, and int64
```

```python
null = (df.isnull().sum())
print("Null values:")
print(null[null>0])
print()
print("Rows and columns of null values:")
print("--------------------------------")
for column in df.columns:
    for row in df.index:
        if pd.isnull(df.loc[row, column]):
            print("Row:", row, "| Column:", column)

print("Columns with null values:")
print("")

# There are two columns with null values.
```

```python
target = "Rating"
features = df.columns.drop([target, "Title"])
print("Target: ", target)
print()
print("Features:\n", features)

# The target is the rating and the useful features are those remaining except the title.
```

```python
print("Missing values:\n")
print((df == 'Unknown').sum())
print()
print()
print("Rows and Columns of missing values:")
print("-----------------------------------")
for column in df.columns:
  for row in df.index:
    if df.loc[row, column] == "Unknown":
      print("Row:",row, "| Column:", column)

# There are missing values.
```

```python

total = df["Rating"].notnull().sum()
print("Movies rated 8.0 or above:", (df["Rating"] >= 8.0).sum())
print("Percentage of movies rated 8.0 or above:", round(((df["Rating"] >= 8.0).sum() / total) * 100, 1), "%")

# You can ask quantitative, qualitative, comparative, and even predictive questions.
# Question: How many movies are rated 8.0 and above and what is the precentage?

```
