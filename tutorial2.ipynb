```python
import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy array :")
print(array_2x2)
```

```python
random_3x4 = np.random.rand(3,4)

print("Random numbers in 3x4 : ")
print(random_3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of zeros : ")
print(zeros_array)

print("Array of ones : ")
print(ones_array)
```

```python
array_6x3 = np.random.rand(6, 3)
print(f"Original : {array_6x3}")

array_6x3_transpose = array_6x3.T
print(f"Transposed :\n{array_6x3_transpose}")
```

```python
n, p, m = 2, 3, 4

array_1 = np.random.rand(n,p)
array_2 = np.random.rand(p,m)
array_3 = np.dot(array_1, array_2)

print(f"First array :\n{array_1} ")
print(f"Second array :\n{array_2}")
print(f"Product :\n{array_3}")
```
